```python
import polars as pl

host = "localhost"
port = 5432
dbname = "db"
user = "peppermint"
password = "peppermint"

uri = f"postgresql://{user}:{password}@{host}:{port}/{dbname}"
```

```python
# polars.read_database_uri uses ADBC/connectorx under the hood — no separate
# sqlalchemy engine needed. Swap the query for whatever table you want to browse.
orders = pl.read_database_uri(
    query="select * from orders limit 1000",
    uri=uri,
)
orders.shape
```

Now run this cell (`:run` / `Ctrl+E`), then either:

- press `gv` to list the kernel's namespace and pick `orders`, or
- run `:view orders` directly

to open it as a read-only grid (`:sql`-style navigation, sort/filter/group with
`gs`/`gf`/`gr`). It's a snapshot — re-run the query cell and `:view` again to refresh.
